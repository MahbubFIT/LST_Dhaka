```python


def gdf_to_fc(gdf):
    features = []
    for _, row in gdf.iterrows():
        geom = ee.Geometry.Polygon(row.geometry.__geo_interface__['coordinates'])
        feat = ee.Feature(geom, row.to_dict())
        features.append(feat)
    return ee.FeatureCollection(features)

dhaka = gdf_to_fc(gdf)
gulshan = dhaka.filter(ee.Filter.eq('NAME_3', 'Gulshan')).first()
gulshan_geom = gulshan.geometry()
def get_premonsoon(year):
    start = ee.Date.fromYMD(year, 3, 1)
    end   = ee.Date.fromYMD(year, 5, 31)
    return start, end
```

```python
def get_LST_image(year):
    start, end = get_premonsoon(year)

    collection = (
        ee.ImageCollection('LANDSAT/LC08/C02/T1_L2')
        .filterBounds(gulshan_geom)
        .filterDate(start, end)
        .sort('CLOUD_COVER')
    )

    img = collection.first()

    # LST band in Kelvin → Celsius
    lst = img.select('ST_B10').multiply(0.00341802).add(149.0).subtract(273.15)
    return lst
```

```python
import ee
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt



# Load shapefile from Google Drive
gdf = gpd.read_file('/content/drive/MyDrive/ARS/Coredhaka.shp')

# Convert GeoDataFrame to Earth Engine FeatureCollection
def gdf_to_fc(gdf):
    features = []
    for _, row in gdf.iterrows():
        # Ensure proper GeoJSON dictionary for ee.Geometry
        ee_geom = ee.Geometry(row.geometry.__geo_interface__)
        feat = ee.Feature(ee_geom, row.to_dict())
        features.append(feat)
    return ee.FeatureCollection(features)

dhaka = gdf_to_fc(gdf)

# Filter Gulshan zone
gulshan = dhaka.filter(ee.Filter.eq('NAME_3', 'Gulshan')).first()

# Assign gulshan_geom directly from the feature's geometry (server-side object)
if gulshan:
    gulshan_geom = gulshan.geometry()
else:
    raise ValueError("Gulshan zone not found in the shapefile.")
```

```python
def get_NDVI_image(year):
    start, end = get_premonsoon(year)

    collection = (
        ee.ImageCollection('LANDSAT/LC08/C02/T1_L2')
        .filterBounds(gulshan_geom)
        .filterDate(start, end)
        .sort('CLOUD_COVER')
    )

    img = collection.first()

    # Red = SR_B4, NIR = SR_B5 (scaled by 0.0000275 + -0.2 normally, but ratio cancels scale)
    red = img.select('SR_B4')
    nir = img.select('SR_B5')
    ndvi = nir.subtract(red).divide(nir.add(red)).rename('NDVI')
    return ndvi
```
